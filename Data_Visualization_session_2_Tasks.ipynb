## Task 5: Detect Possible Outliers

Use these numerical columns:

- `alcohol`
- `malic_acid`
- `color_intensity`
- `proline`

**Requirements:**

1. Create a Box Plot for each feature.
2. Compare the distributions.
3. Calculate Q1 and Q3 for each feature.
4. Calculate the Interquartile Range (IQR).
5. Identify possible outliers using the 1.5 × IQR rule.
6. Count the possible outliers in each feature.
7. Decide whether the outliers should be kept, removed, or investigated.

**Questions:**
- Which features contain possible outliers?
- Does an outlier always mean an incorrect value?
- Why can removing outliers be risky?

**Important:** Do not automatically remove all outliers. Some extreme values may be valid chemical measurements.


```python
features = ["alcohol","malic_acid","color_intensity","proline"]
fig, ax = plt.subplots(1, 4, figsize=(16, 4))
sns.boxplot(data=df, x='alcohol', ax=ax[0])
sns.boxplot(data=df, x='malic_acid', ax=ax[1])
sns.boxplot(data=df, x='color_intensity', ax=ax[2])
sns.boxplot(data=df, x='proline', ax=ax[3])
plt.show()
```

```python
rows = []
for f in features:
    Q1, Q3 = df[f].quantile(0.25), df[f].quantile(0.75)
    IQR = Q3 - Q1
    low, high = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    outliers = df[(df[f] < low) | (df[f] > high)]
    rows.append([f, Q1, Q3, IQR, low, high, len(outliers)])
outlier_table = pd.DataFrame(rows, columns=["feature","Q1","Q3","IQR","lower_bound","upper_bound","n_outliers"])
outlier_table
```

Answers  Task 5
- Decision: keep the outliers
